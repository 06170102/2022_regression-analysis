```python
df3 = pd.read_csv("Problem3a.csv")
```

```python
df3.head()
```

```python
# set x , y
X = df3[['X']]   
y = df3['Y']   

X = sm.add_constant(X) # add constant

# fitting model - OLS linear regression
reg = sm.OLS(y, X).fit()
reg.summary()
```

```python
from bioinfokit.visuz import stat
df3[['yhat']]=pd.DataFrame(reg.predict(X)) #給定X，算出回歸線上 y_hat
df3['res']=pd.DataFrame(reg.resid) #實際y - y_hat 求出殘差
df3['std_res']=reg.get_influence().resid_studentized_internal #計算studentize res
df3.head(2)
```

```python
# 輸出 fitted (y_hat) vs residuals plot 殘差圖。
stat.reg_resid_plot(df=df3, yhat='yhat', resid='res', stdresid='std_res')
```

```python
# Mean 趨近於0
Mean = sum(df3['res'])/len(df3['res'])
Mean
```

```python
import seaborn as sns
#繪製scale location plot，若紅線為水平，代表殘差服從variance condition
p=sns.regplot(x=df3['yhat'], y=np.sqrt(np.abs(df3['std_res'])), lowess=True,line_kws={'color': 'red'})
p.set_title('Scale-Location', fontsize=16)
p.set_xlabel('Fitted Values', fontsize=16)
p.set_ylabel('(Standardized Residuals)^(0.5)', fontsize=16)
```

```python
# 劃出qq- plot，residual不符合常態分佈
sm.qqplot(df3['std_res'], line='45')
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Standardized Residuals")
plt.show()
```

```python
influence = reg.get_influence()
inf_sum = influence.summary_frame()
inf_sum.head()

```

```python
inf_sum[inf_sum['cooks_d'] >1] # no obs cooks distance > 1
```

```python
d = 2 / (100**0.5)
```

```python
inf_sum[abs(inf_sum['dfb_X']) > d] # no obs dfbx > d
```

```python
#4d  using Least absolute error instead
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
mod = smf.quantreg("Y ~ X", df7) #採用 quantile regression model 
res = mod.fit(q=0.5) # q= 0.5 將跟LAD等價
print(res.summary())
```

```python
#根據LAD法，估出迴歸係數。 bo_hat =  0.506493 b1_hat = 0.696224。 斜率較ＯＬＳ法大，受離群值影響較小。
```

```python
#將結果畫在同張圖上比較。紅線為LAD迴歸線，藍線為OLS迴歸線。可看出ols受離群值影響，導致線被往下帶，較為平。
#而LAD為受影響較小。
x = np.arange(0, 50)


ax = df7.plot(x='X', y='Y', kind='scatter')
ax.plot(x, p2.Intercept + p2.X * x,color = 'r')# LAD line  Y = 0.506493 + 0.696224 X 
ax.plot(x, p.params.const + p.params.X * x) #ols line Y = 4.4475 + 0.4622 * X

```

```python
# set x , y
X = df7[['X']]   
y = df7['Y']   

X = sm.add_constant(X) # add constant

# 採用 statsmodel robust_linear_model
#Estimate a robust linear model via iteratively reweighted least squares given a robust criterion estimator.
rlm = sm.RLM(y, X).fit()
print(rlm.summary())
```

```python
print(rlm.params)
```

```python
# RLM 和 LAD 迴歸線相差不多，導致兩線重疊，無法辨識。
x = np.arange(0, 50)
ax = df7.plot(x='X', y='Y', kind='scatter')
ax.plot(x, p2.Intercept + p2.X * x,color = 'r') # LAD line  Y = 0.506493 + 0.696224 X 
ax.plot(x, p.params.const + p.params.X * x,color = 'b') #ols line Y = 4.4475 + 0.4622 * X
ax.plot(x,rlm.params.const + rlm.params.X * x,color = 'g') #RLM Y = 0.4739 + 0.6969 * X
plt.show()
```
