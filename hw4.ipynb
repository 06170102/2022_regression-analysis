```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Table01.csv') #讀取資料
```

```python
df.head() # check top5 rows
```

```python
df.drop(df.columns[[0, 1]], axis=1, inplace=True) # drop first 2 columns, not needed
```

```python
df= df.replace('-', 0) # 用0補值，每筆data是和前個月相差， - 表示兩月數值相同，無變動
```

```python
df
```

```python
df.dtypes #check dtypes
```

```python
df= df.astype(float) # convert to float
```

```python
df.dtypes
```

```python
# rename columns
df.columns = ['salary', 'recurring','non_recurring','overtime_pay','average_hour','normal_hour','overtime_hour','enter','exit','wage_index','recurring_index']
```

```python
df.describe()
```

```python
#1. ============================================================================
```

```python
y = df['exit'] 
X = df.loc[:, df.columns != 'exit']
```

```python
import statsmodels.api as sm
X = sm.add_constant(X)   #增加模型的常數

model0 = sm.OLS(y,X)    #OLS
results0 = model0.fit() #fitting

print(results0.summary())
```
